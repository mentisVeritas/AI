# Kirish

"Avtomobil savdosi ma’lumotlari" to‘plami bir yil davomida avtosalondagi avtomobil savdolari haqidagi keng qamrovli ma’lumotlar to‘plamidir. Ushbu to‘plam har bir avtomobil savdosi uchun 2,5 million qator va to‘qqizta ustundan iborat. To‘plamda sotuv sanasi, sotuvchi nomi, xaridor ismi, avtomobil markasi va modeli, ishlab chiqarilgan yili, AQSH dollaridagi sotuv narxi, sotuvchiga to‘lanadigan komissiya stavkasi hamda sotuvchi olgan komissiya miqdori kabi batafsil ma’lumotlar mavjud.

Ushbu ma’lumotlar to‘plami vaqt o‘tishi bilan avtomobil savdosi tendensiyalarini tahlil qilish, avtosalondagi alohida sotuvchilarning faoliyatini baholash va savdo samaradorligini bashorat qilishda foydali manba hisoblanadi. Shuningdek, u avtomobil markasi, modeli, ishlab chiqarilgan yili va sotuvchi faoliyati kabi turli omillarning avtomobil savdosiga ta’sirini o‘rganish uchun ham qo‘llanilishi mumkin. Bu to‘plam avtomobil savdosi sohasiga qiziquvchilar yoki savdo ma’lumotlarini tahlil qilmoqchi bo‘lganlar uchun ajoyib manbadur.

```python
unique_car_makes = df['Car Make'].unique()
unique_car_models = df['Car Model'].unique()

print("Unique Car Makes: \n", unique_car_makes)
print("Unique Car Models: \n", unique_car_models)
```

```python
car_make_counts = df['Car Make'].value_counts()
print(car_make_counts)
```

```python
car_model_perc = df['Car Model'].value_counts(normalize=True) * 100
print(car_model_perc)

car_make_perc = df['Car Make'].value_counts(normalize=True) * 100
print(car_make_perc)
```

```python
import matplotlib.pyplot as plt

# Get value counts for car makes
car_make_counts = df['Car Make'].value_counts()

# Create a pie chart
fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(car_make_counts, labels=car_make_counts.index, autopct='%1.1f%%', startangle=90)
ax.axis('equal')
ax.set_title('Distribution of Car Makes')

plt.show()
```

**Top-5 avtomobil modellari Silverado, Civic, Corolla, F-150 va Altima bo‘lib, ularning foiz taqsimoti mos ravishda 20,03%, 20,02%, 20,01%, 19,99% va 19,95% ni tashkil etadi.**

```python
import matplotlib.pyplot as plt

# Get value counts for car makes
car_make_counts = df['Car Model'].value_counts()

# Create a pie chart
fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(car_make_counts, labels=car_make_counts.index, autopct='%1.1f%%', startangle=90)
ax.axis('equal')
ax.set_title('Distribution of Car Models')

plt.show()
```

**Eng yuqori 5 ta avtomobil markasi Honda, Chevrolet, Toyota, Ford va Nissan bo‘lib, ularning foiz taqsimoti tegishlicha 20,03%, 20,02%, 20,01%, 19,99% va 19,96% ni tashkil etadi.**

**Yearly Sales**

```python
yearly_sales = df.groupby('Car Year')['Sale Price'].sum().reset_index()
yearly_sales = yearly_sales.sort_values(by='Sale Price', ascending=False)
print(yearly_sales)
```

```python
import matplotlib.pyplot as plt

# Group the data by year and calculate the total sales for each year
yearly_sales = df.groupby('Car Year')['Sale Price'].sum()

# Calculate the percentage of sales for each year
yearly_sales_perc = yearly_sales / yearly_sales.sum() * 100

# Create a line chart to visualize the yearly sales percentage
plt.plot(yearly_sales_perc.index, yearly_sales_perc.values)
plt.xlabel('Year')
plt.ylabel('Percentage of Sales')
plt.title('Yearly Sales Percentage')
plt.show()
```

**2015-yil eng yuqori savdo ko‘rsatkichiga ega bo‘lgan, 2022-yil esa eng past savdo ko‘rsatkichiga ega bo‘lgan yil hisoblanadi.**

**Turli avtomobil modellari o‘rtasida olinadigan vositachilik haqi miqdorida farq mavjudmi?**

```python
model_commission = df.groupby('Car Model')['Commission Earned'].mean()
print(model_commission)
```

**model_commission** qiymatlari har bir avtomobil modeli uchun o‘rtacha olingan komissiya miqdori nisbatan o‘xshash ekanligini ko‘rsatadi, bunda qiymatlar **2997.87 dan 3003.31** gacha oraliqda. Biroq, bu farqlar juda kichik va amalda muhim ahamiyatga ega bo‘lmasligi mumkin.

**Ba’zi sotuvchilar sotish hajmi yoki yig‘ilgan komissiya miqdori bo‘yicha boshqalardan ko‘ra samaraliroq faoliyat ko‘rsatadimi?**

```python
salesperson_sales = df.groupby('Salesperson')['Sale Price'].sum()
salesperson_commission = df.groupby('Salesperson')['Commission Earned'].sum()

# Sort by total sales
salesperson_sales = salesperson_sales.sort_values(ascending=False)

# Sort by commission earned
salesperson_commission = salesperson_commission.sort_values(ascending=False)
```
