```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from apyori import apriori
```

## Data Preprocessing

```python
dataset = pd.read_csv('../../data/Market_Basket_Optimisation.csv', header=None)
```

```python
dataset
```

```python
transactions = []

for i in range(0, 7501):
    transactions.append([str(dataset.values[i, j]) for j in range(0, 20) if str(dataset.values[i, j]) != 'nan'])
```

## Training the Apriori model on the dataset

```python
rules = apriori(transactions=transactions, min_support=0.01, min_confidence=0.3, min_lift=1.01)
```

## Visualising the results

### Displaying the first results coming directly from the output of the apriori function

```python
results = list(rules)
```

### Putting the results well organised into a Pandas DataFrame

```python
res = results[1]
res.ordered_statistics.items_base
```

```python
def inspect(results):
    lhs = [tuple(result[2][0][0])[0] for result in results]
    rhs = [tuple(result[2][0][1])[0] for result in results]
    supports = [result[1] for result in results]
    confidences = [result[2][0][2] for result in results]
    lifts = [result[2][0][3] for result in results]
    return list(zip(lhs, rhs, supports, confidences, lifts))


resultsinDataFrame = pd.DataFrame(inspect(results),
                                  columns=['Left Hand Side', 'Right Hand Side', 'Support', 'Confidence', 'Lift'])
```
