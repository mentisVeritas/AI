```python
from statistics import median

import pandas as pd
import numpy as np
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
df.sample(11)
```

### Bir xil malumotli ustunlar:
1. survived & alive, biz alive tushurib qoldiramiz,sabab u kategorikal
2. pclass va class : class ni tushirib qoldiramiz, sababi u kategorikal
3. sex,who,adult_male: 2tasini tushirib qoldiramiz

```python
df.info()
```

```python
df = df.drop(columns=['alive', 'class', 'sex', 'adult_male'])
df.head()
```

```python
df.isna().sum()
```

```python
df.isna().sum()/len(df)*100
```

```python
#Возраст запольняем с медианой
median = df['age'].median()
df['age'] = df['age'].fillna(median)

#Город заполняем с модой потому что категорикал
moda = df['embark_town'].mode()
df['embark_town']=df['embark_town'].fillna(moda)
```

```python
df.groupby(['embark_town','embarked']).count()
```

```python
# Выбрасываем Embarked так как оно содержить меньше информацый чем Embarked town
df = df.drop(columns=['embarked'])
```

```python
df.head()
```

```python
df = df.drop(columns=['deck'])
```

```python
df.head()
```

```python
moda = df['embark_town'].mode()
df['embark_town'] = df['embark_town'].fillna(moda[0])
df.isna().sum()
```

```python
df['who'] = df['who'].replace({'man':1, 'woman':0})
df['alone'] = df['alone'].replace({True:1, False:0})
```

```python
df.head()
```

### Embark_town ustunini on-hot encoding orqali numerical ga otkazamiz, buning uchun pd.get_dummies()


```python
new_collums = ['Cherbourg','Queenstown','Southampton']
df[new_collums] = pd.get_dummies(df['embark_town'],dtype=int)
df = df.drop(columns=['embark_town'])
df.head()
```
