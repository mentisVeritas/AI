## 1️⃣ Pandas `Series` nima?
**Series** — bu Pandas kutubxonasidagi **bir o‘lchovli** (1D) ma’lumotlar tuzilmasi bo‘lib, u indekslangan qiymatlar qatoridan iborat.

**Xususiyatlari:**
- **O‘ziga xos indeks**: Har bir element maxsus indeksga ega.
- **Homogen ma’lumotlar**: Asosan bir turdagi ma’lumotlarni saqlaydi (int, float, string va h.k.).
- **Ko‘rinishi**: `numpy array`ga o‘xshash.

### ✅ `Series` misoli:
```python
import pandas as pd

# Oddiy Series yaratish
s = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print(s)

````python
a = [1, 7, 2]

myvar = pd.Series(a)

print(myvar)
````

````python
a = [1, 7, 2]

myvar = pd.Series(a, index = ["x", "y", "z"])

print(myvar)
````

````python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

#load data into a DataFrame object:
df = pd.DataFrame(data, columns=["calories", "duration"],index = ["x", "y", "z"])
print(df)
````

````python
calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories,index = ["day1","day2", "day3"],name="calories")
myvar
````
