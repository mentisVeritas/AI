```python
arr1=np.arange(12)
print(arr1)
print(arr1.ndim)
print(arr1.size)
print(arr1.shape)
```

2 dimensional array=>Matrix

```python
arr2=arr1.reshape(3,4)
print(arr2)
print(arr2.ndim)
print(arr2.size)
print(arr2.shape)
```

```python
arr3=arr2.T
print(arr3)
print(arr3.ndim)
print(arr3.size)
print(arr3.shape)
```

```python
arr4=np.arange(3*3*3).reshape(3,3,3)
print(arr4)
print(arr4.ndim)
print(arr4.size)
print(arr4.shape)
type(arr4)
print(arr4.dtype)
```
